**Paso 1: Limpiar los datos**

```python
import pandas as pd

url = "https://github.com/DavidCruz1013/Canasta-Basica-en-El-Salvador-/raw/refs/heads/main/datos/CANASTA%20B%C3%81SICA%20ALIMENTARIA.xlsx"

# Leer archivo
df = pd.read_excel(url)

# Ver primeras filas
df.head()
```

**Paso 2: Corregir nombres de columnas**

```python
# Tomar la fila 0 como encabezados
df.columns = ['AÑO', 'ARTICULO', 'GRAMOS_PERSONA',
              'ENERO', 'FEBRERO', 'MARZO', 'ABRIL',
              'MAYO', 'JUNIO', 'JULIO', 'AGOSTO',
              'SEPTIEMBRE', 'OCTUBRE', 'NOVIEMBRE', 'DICIEMBRE']

# Eliminar filas basura
df = df.iloc[2:]

# Reiniciar índice
df.reset_index(drop=True, inplace=True)

# Mostrar datos limpios
df.head()
```

**Convertir columnas a numéricas**

```python
meses = ['ENERO','FEBRERO','MARZO','ABRIL',
         'MAYO','JUNIO','JULIO','AGOSTO',
         'SEPTIEMBRE','OCTUBRE','NOVIEMBRE','DICIEMBRE']

for mes in meses:
    df[mes] = pd.to_numeric(df[mes], errors='coerce')
```

**Promedio Mensual de la canasta Basica **

```python
import matplotlib.pyplot as plt

df[meses].mean().plot(kind='bar')

plt.title("Promedio mensual de la canasta básica")
plt.xlabel("Meses")
plt.ylabel("Precio promedio")

plt.show()
```

**Productos con mayor precio promedio en 2020**

```python
meses = ['ENERO','FEBRERO','MARZO','ABRIL',
         'MAYO','JUNIO','JULIO','AGOSTO',
         'SEPTIEMBRE','OCTUBRE','NOVIEMBRE','DICIEMBRE']

# Filtrar año 2020
df_2020 = df[df['AÑO'] == 2020]

# Promedio anual por producto
df_2020['PROMEDIO'] = df_2020[meses].mean(axis=1)

# Top productos más caros
top_productos = df_2020[['ARTICULO','PROMEDIO']].sort_values(by='PROMEDIO', ascending=False)

top_productos.head(10)
```

**Gráfica de productos más caros**

```python
import matplotlib.pyplot as plt

top10 = top_productos.head(10)

plt.figure(figsize=(10,5))

plt.bar(top10['ARTICULO'], top10['PROMEDIO'])

plt.xticks(rotation=45)

plt.title("Productos con mayor precio promedio en 2020")
plt.xlabel("Producto")
plt.ylabel("Precio promedio")

plt.show()
```

**¿Qué productos desaparecieron?**

Algunos productos presentan períodos con valores iguales a cero, lo que puede indicar ausencia temporal del producto en la canasta básica o falta de registros durante ciertos meses

```python
df['CEROS'] = (df[meses] == 0).sum(axis=1)

productos_desaparecidos = df[['ARTICULO','CEROS']].sort_values(by='CEROS', ascending=False)

productos_desaparecidos.head(10)
```

**Cuánto aumentó la canasta básica por año**

```python
# Promedio anual
promedio_anual = df.groupby('AÑO')[meses].mean().mean(axis=1)

promedio_anual
```

**Gráfica de incremento anual**

```python
plt.figure(figsize=(8,5))

promedio_anual.plot(marker='o')

plt.title("Incremento promedio de la canasta básica")
plt.xlabel("Año")
plt.ylabel("Precio promedio")

plt.grid(True)

plt.show()
```

**Año con tasa más baja**

```python
promedio_anual.idxmin()
```

**Año con tasa más alta**

```python
promedio_anual.idxmax()
```

**6. Producto más estable**

```python
df['VARIACION'] = df[meses].std(axis=1)

productos_estables = df[['ARTICULO','VARIACION']].sort_values(by='VARIACION')

productos_estables.head(10)
```

**Producto con mayor inflación**

```python
productos_inflacion = df[['ARTICULO','VARIACION']].sort_values(by='VARIACION', ascending=False)

productos_inflacion.head(10)
```

```python
df_largo = df.melt(
    id_vars=['AÑO', 'ARTICULO', 'GRAMOS_PERSONA'],
    value_vars=meses,
    var_name='MES',
    value_name='PRECIO'
)

df_largo.head()
```

AGREGAR ZONA EN EL NUEVO CSV

```python
# Crear columna zona
df["ZONA"] = ""

df.loc[0:111, "ZONA"] = "Urbana"
df.loc[116:, "ZONA"] = "Rural"

df = df.drop([113,114,115])

df.reset_index(drop=True, inplace=True)
```

LIMPIEZA ENCABEZADOS

```python
# Eliminar filas completamente vacías
df = df.dropna(how='all')

# Eliminar fila que repite encabezados
df = df[df['AÑO'] != 'AÑO']

# Reiniciar índices
df = df.reset_index(drop=True)

```

**Descargar CSV LIMPIO**

```python
df.to_csv(
    'canasta_limpia.csv',
    index=False,
    encoding='utf-8-sig'
)

from google.colab import files
files.download('canasta_limpia.csv')
```
